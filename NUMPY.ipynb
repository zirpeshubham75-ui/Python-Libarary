# 1. What is NumPy?

Answer:

NumPy (Numerical Python) is a Python library used for numerical computing. It provides a powerful N-dimensional array object and mathematical functions  on large datasets.



# 2. Why is NumPy faster than Python lists?

Answer:

NumPy is faster because:

It stores elements in contiguous memory.
All elements have the same data type.
It performs operations using optimized C code.
It supports vectorized operations, avoiding Python loops.

# 3 Difference between Python List and NumPy Array
Python List	                      NumPy Array
Can store different data types	 Stores same data type
Slower	                         Faster
More memory usage	               Less memory usage
Doesn't support vectorized operations	Supports vectorized operations


# 4  What is Vectorization?

Answer:

Vectorization means performing operations on an entire array without writing loops.

 # reshape()

# HANDLIN THE MISSING VALUES
 - np.isnan     detect missing values
 - np.nan_to_num()       # not a number 
 - np.isinf      # to dectect the infinte value 

```python
a=np.array([1,2,np.nan,4,np.nan,6])
np.isnan(a)
```

```python
       #  np.nan_to_num()  
```

```python
a=np.array([1,2,np.nan,4,np.nan,6])
ca = np.nan_to_num(a)          # default value is 0
ca
```

```python
ca = np.nan_to_num(a,nan=15)
ca
```

```python
a = np.array([1,2,np.inf,4,-np.inf,6])
np.isinf(a)
```

```python
np.nan_to_num(a,posinf=1000, neginf=-1000) 
```
