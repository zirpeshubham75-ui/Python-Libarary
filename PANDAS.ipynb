# What is Data Manipulation and Analysis?
# Data Manipulation
Definition: Changing, organizing, or preparing data to make it useful and easier to understand.
Goal: To clean, transform, and structure raw data for better usability.
Example:
Organizing a Grocery List: Sorting random items into categories like "Fruits" or "Dairy".
Fixing Errors in a Student Record: Correcting missing or wrong grades.
# Data Analysis
Definition: Extracting patterns, trends, and insights from the data to solve problems.
Goal: To answer questions or identify trends using the data.
Example:
Analyzing Sales Trends: Finding the month with the highest revenue.
Tracking Fitness Progress: Analyzing daily steps and calories for activity.

# FILLING PERTIUCLA NAN VALUE  

1 interploate method 
  - when to use
    1  timer series data
    2  numeric data with follow same treand
    3 avoid dropping rows



    # linear interploate = straight-line estimation.
# polynomial interploate = curve-based estimation.
# time interploate = interpolation based on timestamps.

```python
df = pd.DataFrame({
    "Date": pd.to_datetime([
        "2025-01-01",
        "2025-01-02",
        "2025-01-03",
        "2025-01-04",
        "2025-01-05",
        "2025-01-06",
        "2025-01-07"
    ]),
    
    # Straight trend
    "Linear": [10, 20, np.nan, 40, 50, np.nan, 70],
    
    # Curved trend (approximately x²)
    "Polynomial": [1, 4, np.nan, 16, 25, np.nan, 49],
    
    # Time series
    "Sales": [100, np.nan, 140, np.nan, 180, 200, np.nan]
})

df
```

```python
df["Linear"] = df["Linear"].interpolate(method="linear")
df
```

# What is order?

order tells Pandas what degree of polynomial (curve) to use when estimating missing values.

Order	Polynomial Type	Equation
# 1	Linear	y=ax+b
# 2	Quadratic	y=ax

+bx+c
# 3	Cubic	y=ax

```python
df["Polynomial"] = df["Polynomial"].interpolate(method="polynomial", order=2)
df
```

```python
df = df.set_index("Date") # Set Date as the index
df["Sales"] = df["Sales"].interpolate(method="time")
df
```

# MERGE AND JOIN THE DATA SET

```python
df_customers = pd.DataFrame ({

   'CustomerID' : [1,2,3],
    'Name'    : ['shubham','suresh','kalpesh']

})



df_Orders = pd.DataFrame ({

  'CustomerID'  : [1,2,4],
    'Orderamout' :[250,350,450]

})
df_customers

    
```

# inner join 

```python
df_merge = pd.merge(df_customers,df_Orders,on='CustomerID',how="inner")
df_merge
```

# outer join

```python
df_merge = pd.merge(df_customers,df_Orders,on='CustomerID',how="outer")
df_merge
```

# LEFT JOIN

```python
df_merge = pd.merge(df_customers,df_Orders,on='CustomerID',how="left")
df_merge
```

# right join  

```python
df_merge = pd.merge(df_customers,df_Orders,on='CustomerID',how="right")
df_merge
```
